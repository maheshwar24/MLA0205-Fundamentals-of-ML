```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

X = np.array([
    [1, 10000, 100], [2, 12000, 100],
    [3, 11000, 95], [4, 13000, 95],
    [5, 15000, 90], [6, 14000, 90],
    [7, 16000, 90], [8, 17000, 85],
    [9, 18000, 85], [10, 20000, 80],
    [11, 19000, 80], [12, 22000, 80],
    [13, 21000, 75], [14, 23000, 75],
    [15, 25000, 75], [16, 24000, 70],
    [17, 26000, 70], [18, 28000, 70],
    [19, 27000, 65], [20, 30000, 65],
    [21, 29000, 65], [22, 32000, 60],
    [23, 31000, 60], [24, 34000, 60]
])

y = np.array([
    120, 135, 130, 145, 160, 155,
    170, 180, 190, 210, 205, 230,
    225, 240, 260, 255, 275, 290,
    285, 310, 305, 330, 325, 350
])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

print("Future Sales Prediction")
print("Actual Sales:", y)
print("Predicted Sales:", np.round(y_pred, 2))
print("Mean Squared Error:", mean_squared_error(y, y_pred))
print("R2 Score:", r2_score(y, y_pred))

next_month = np.array([[25, 35000, 60]])
future_sales = model.predict(next_month)

print("Predicted Sales for Month 25:", round(future_sales[0], 2))
```
